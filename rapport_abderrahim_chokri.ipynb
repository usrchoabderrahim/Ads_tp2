## Introduction

Ce travail etudie la formation des opinions avec une simulation multi-agents realisee avec Mesa 3. Chaque agent represente une personne. Son opinion est un nombre entre 0 et 1. e chaque etape, il observe ses voisins et rapproche son opinion de la moyenne de leurs opinions.

Les simulations utilisent une grille de 10 e 10 agents et 50 etapes. Les graphiques enregistres dans le notebook sont presentes ci-dessous pour trois valeurs de `alpha` : 0, 0,5 et 1.

## Resultats des simulations

### `alpha = 0,5`

La variance diminue rapidement, puis se stabilise pres de zero. Les opinions des agents se rapprochent donc les unes des autres. La moyenne reste presque stable.

Les trajectoires individuelles montrent aussi que les opinions se rapprochent progressivement.

### `alpha = 0`

Les agents ne changent pas leur opinion. La moyenne et la variance restent constantes, et les trajectoires individuelles sont presque horizontales.

### `alpha = 1`

Les agents prennent directement leopinion moyenne de leurs voisins. La variance baisse tres rapidement dans cette simulation. La moyenne globale reste presque stable.

## Reponses aux cinq premieres questions de la derniere cellule

### 1. Quel est leimpact de `alpha` sur la convergence e

`alpha` indique la force de leinfluence des voisins. Quand `alpha` est proche de 0, chaque agent change tres peu son opinion. La convergence est donc lente, et avec `alpha = 0`, les opinions ne changent pas. Quand `alpha = 1`, leagent adopte entierement la moyenne des opinions de ses voisins. Le rapprochement est plus rapide. Avec `alpha = 0,5`, le changement est intermediaire et les opinions se rapprochent progressivement.

### 2. Pourquoi la polarisation est-elle difficile dans cette configuration e

e chaque etape, leagent se deplace vers la moyenne des opinions voisines. Cette moyenne se trouve entre les opinions observees. Leagent ne seeloigne donc pas de ses voisins e cause de cette regle. Comme les agents sont connectes sur la grille et repetent ce rapprochement, les differences diminuent et les opinions finissent par devenir proches. Le modele ne contient pas de regle qui renforce les desaccords ou qui separe les agents en groupes.

### 3. Comment leemergence apparaet-elle dans les graphiques e

Leemergence signifie queun resultat collectif apparaet e partir de regles simples appliquees par chaque agent. Ici, chaque personne suit seulement leopinion de ses voisins. Pourtant, la variance diminue e leechelle de toute la population et les trajectoires se rapprochent. Le consensus est donc un resultat collectif, sans chef qui donne une opinion commune.

### 4. Quel est le rele de `self.agents.shuffle_do("step")` e

Cette instruction demande e Mesa deappeler la methode `step()` de tous les agents dans un ordre melange. Une boucle classique sur une liste suit generalement leordre des elements. Leordre aleatoire evite de toujours donner la priorite aux memes agents. Il rend aussi la simulation plus representative deinteractions qui ne se passent pas toujours dans le meme ordre.

### 5. Quelle est la difference entre `model_reporters` et `agent_reporters` e

`model_reporters` enregistrent des mesures de toute la population, comme leopinion moyenne et la variance. `agent_reporters` enregistrent une mesure pour chaque agent, ici son opinion. Les deux sont utiles : les mesures globales montrent si la population converge et si sa dispersion diminue ; les mesures individuelles permettent de voir les trajectoires et les differences entre agents.
